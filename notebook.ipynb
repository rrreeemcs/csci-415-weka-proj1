```python
import pandas as pd

df = pd.read_csv('data/Algerian_forest_fires_dataset_UPDATE.csv', header=1)
df.head()
```

```python
# Splitting df into two parts (Bejala & Sidi-Bel Abbes Regions)
# Creating new 'region' column for each and setting as index
bejala_df = df.iloc[:122]
bejala_df['region'] = 'Bejala'
bejala_df.set_index('region', inplace=True)

sidi_bel_abbes_df = df.iloc[124:]
sidi_bel_abbes_df['region'] = 'Sidi-Bel Abbes'
sidi_bel_abbes_df.set_index('region', inplace=True)

# Combining both parts into full algerian_forest_fires DataFrame
algerian_forest_fires_df = pd.concat([bejala_df, sidi_bel_abbes_df], axis=0)
```

```python
# Viewing the Bejala region of the combined df
algerian_forest_fires_df.loc['Bejala']
```

```python
# Viewing the Sidi-Bel Abbes region of the combined df
algerian_forest_fires_df.loc['Sidi-Bel Abbes']
```

```python
# Viewing shape of algerian fires Df
print(f"Number of Instances: {algerian_forest_fires_df.shape[0]}")
print(f"Number of Attributes: {algerian_forest_fires_df.shape[1]}")
```

```python
# Stripping away any whitespace from columns
algerian_forest_fires_df.columns = algerian_forest_fires_df.columns.str.strip()
```

```python
# Viewing data types of all columns -> change Temperature, RH, Ws, and Rain to floats
print(algerian_forest_fires_df.dtypes)

algerian_forest_fires_df.astype({
    'Temperature': 'float',
    'RH': 'float',
    'Ws': 'float',
    'Rain': 'float',
})
```
