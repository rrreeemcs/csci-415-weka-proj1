```python
# Reading the Algerian Forest Fires Dataset (right from the link)
import pandas as pd

df = pd.read_csv('data/Algerian_forest_fires_dataset_UPDATE.csv', header=1)
df.head()
```

#### Preprocess 1
- Dataset contains TWO tables, each for different region (Bejala & Sidi-Bel Abbes)
- Using .iloc to split the dataset into its respective parts and set their new index to be the region name
- Combining the separate DataFrames together to make algerian_ff

```python
# Bejala region index 0:122 in original df
bejala_df = df.iloc[:122]
bejala_df['region'] = 'Bejala'
bejala_df.set_index('region', inplace=True)

# Sidi-Bel Abbes region index 124:end in original df
sidi_bel_abbes_df = df.iloc[124:]
sidi_bel_abbes_df['region'] = 'Sidi-Bel Abbes'
sidi_bel_abbes_df.set_index('region', inplace=True)

# Combining both parts into algerian_ff
algerian_ff = pd.concat([bejala_df, sidi_bel_abbes_df], axis=0)
```

```python
# Viewing the full Bejala region information
algerian_ff.loc['Bejala']
```

```python
# Viewing the full Sidi-Bel Abbes region information
algerian_ff.loc['Sidi-Bel Abbes']
```

```python
# Viewing shape of algerian fires Df
print(f"Number of Instances: {algerian_ff.shape[0]}")
print(f"Number of Attributes: {algerian_ff.shape[1]}")
```

#### Preprocess 2
Focusing on the column formatting of the algerian_ff DataFrame
- CSV showed there was whitespace visible for some column names -> stripping the whitespace away
- CSV also showed there was some whitespace from some fields -> stripping the whitespace away
- Changing the columns to lowercase for readability
- Converting the numerical values to floats (FFMC:FWI are index codes and should not be changed)
- Renaming some columns to include their unit of measurement (Temperature -> temp_c)

```python
# Stripping away the whitespace from the columns & fields
algerian_ff.columns = algerian_ff.columns.str.strip()
algerian_ff = algerian_ff.apply(lambda x: x.str.strip() if x.dtype == "str" else x)

# Changing the column names to lowercase
algerian_ff.columns = algerian_ff.columns.str.lower()
```

```python
# Viewing data types of all columns -> change Temperature, RH, Ws, and Rain to floats
algerian_ff.dtypes
```

```python
# Changing temperature, rh, ws, rain to float datatypes
algerian_ff = algerian_ff.astype({
    'temperature': 'float',
    'rh': 'float',
    'ws': 'float',
    'rain': 'float',
})

# Renaming columns with changed datatypes to match the unit of measure + lowercase
algerian_ff.rename({
    'temperature':'temp_c',
    'rh':'rel_humidity_percent',
    'ws':'wind_speed_kmh',
    'rain':'rainfall_mm',
}, axis=1, inplace=True)

# Checking to see if the data types of the columns changed
algerian_ff.dtypes
```

#### Preprocess 3
- Creating a new date column (potentially for line graph analysis) and move to index 0
- Removing any rows containing null class values

```python
# Creating date column to combine day-month-year together
algerian_ff['date'] = pd.to_datetime(algerian_ff[['year', 'month', 'day']])
date_col = algerian_ff.pop('date')
algerian_ff.insert(0, 'date', date_col)

# Checking if column moved properly
algerian_ff.head()
```

```python
# Finding any NA values from Df -> 1 null class meaning we will just remove that row
algerian_ff.isnull().sum()
```

```python
# Dropping the row with the NA class value
algerian_ff.dropna(axis=0, subset='classes', inplace=True)

# Verifying if all null values are gone
algerian_ff.isnull().sum() > 0
```

### Final Step
Converting newly cleaned algerian_ff to clean CSV in data directory
